```python
def count_with_status(orders, status):
    count = 0
    for order in orders:
        if order["status"] == status:
            count += 1
    return count


def average_value(orders, status):
    total = total_value(orders, status)
    count = count_with_status(orders, status)
    
        
    return round(total / count, 2)
```

```python
def total_per_status(orders):
    totals = {}
    for order in orders:
        status = order["status"]
        value = order["value"]
        
        if status not in totals:
            totals[status] = 0.0
            
        totals[status] += value
        
    for status in totals:
        totals[status] = round(totals[status], 2)
        
    return totals
```

```python
def total_by(rows, group_key, value_key):
    totals = {}
    for row in rows:
        group = row[group_key]
        value = row[value_key]
        
        if group not in totals:
            totals[group] = 0.0
            
        totals[group] += value
        

    for group in totals:
        totals[group] = round(totals[group], 2)
        
    return totals
```
