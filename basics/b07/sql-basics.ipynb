```python
duckdb.sql("""
       SELECT
           COUNT(*)            AS payments,
           SUM(payment_value)  AS total_paid,
           AVG(payment_value)  AS average_payment,
           MIN(payment_value)  AS smallest,
           MAX(payment_value)  AS biggest
       FROM payments
   """)
```

```python
duckdb.sql("SELECT ROUND(SUM(payment_value), 2) AS total_paid FROM payments")
```

```python
   duckdb.sql("""
       SELECT
           COUNT(*)                              AS all_orders,
           COUNT(order_delivered_customer_date)  AS with_delivery_date,
           COUNT(order_approved_at)              AS with_approval_date
       FROM orders
   """)
```

```python
   duckdb.sql("""
       SELECT strftime(order_purchase_timestamp, '%Y-%m') AS month, COUNT(*) AS orders
       FROM orders
       GROUP BY month
       ORDER BY month
   """)
   
```

```python
   duckdb.sql("""
       SELECT strftime(order_purchase_timestamp, '%Y-%m') AS month, COUNT(*) AS orders
       FROM orders
       GROUP BY month
       HAVING COUNT(*) > 7000
       ORDER BY month
   """)
```
