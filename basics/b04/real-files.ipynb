```python
def total_by(rows, group_key, value_key):
    totals = {}
    for row in rows:
        group = row[group_key]
        value = row[value_key]
        
        if group not in totals:
            totals[group] = 0.0
            
        totals[group] += value
        

    for group in totals:
        totals[group] = round(totals[group], 2)
        
    return totals
```
